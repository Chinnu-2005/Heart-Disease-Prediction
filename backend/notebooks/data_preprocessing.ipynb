```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv('../data/heart.csv')
```

```python
target="HeartDisease"
```

```python
X=data.drop(columns=[target])
y=data[target]
```

```python
X["RestingBP"]=X["RestingBP"].replace(0,np.nan)
X["Cholesterol"]=X["Cholesterol"].replace(0,np.nan)
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(
    X,y,test_size=0.3,random_state=40,stratify=y
)
```

```python
import sklearn.impute as impute
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,OneHotEncoder
```

```python
numerical_features=["Age","RestingBP","Cholesterol","MaxHR","Oldpeak"]

categorical_features=["Sex",
    "ChestPainType",
    "FastingBS",
    "RestingECG",
    "ExerciseAngina",
    "ST_Slope"
]


num_transformer=Pipeline(
    [
        ("imputer",impute.SimpleImputer(strategy="median")),
        ("scaler",StandardScaler())
    ]
)

cat_transformer=Pipeline(
    [
        ("imputer",impute.SimpleImputer(strategy="most_frequent")),
        ("encoder",OneHotEncoder(handle_unknown="ignore"))
    ]
)
```

```python
from sklearn.compose import ColumnTransformer

preprocessor=ColumnTransformer(
    [
        ("num",num_transformer,numerical_features),
        ("cat",cat_transformer,categorical_features)
    ]
)

```

```python
X_train_preprocessed=preprocessor.fit_transform(x_train)
X_test_preprocessed=preprocessor.transform(x_test)

print("X_train_preprocessed shape:",X_train_preprocessed.shape)
print("X_test_preprocessed shape:",X_test_preprocessed.shape)
```
