```python
print(df.groupby(df['Cholesterol'].eq(0))['HeartDisease'].value_counts(normalize=True))
```

```python
print(df.groupby(
    df['Cholesterol'].eq(0)
).agg(
    count=('Cholesterol', 'size'),
    heart_disease_rate=('HeartDisease','mean'),
    avg_age=('Age','mean')
)
)
```

```python
pd.crosstab(
    df['Cholesterol'].eq(0),
    df['HeartDisease']
)
```

```python
print(
    df.groupby(df["Cholesterol"].eq(0))[
        ["Age", "RestingBP", "MaxHR", "Oldpeak"]
    ].mean()
)
```

```python
print(
    df.groupby("HeartDisease")["Cholesterol"]
      .apply(lambda x: (x == 0).mean())
)
```

```python
for col in ["Age","Sex","ChestPainType","ExerciseAngina","ST_Slope"]:
    print("----------------------------")
    print(
        pd.crosstab(
            df['Cholesterol'].eq(0),
            df[col],
            normalize="index"
        ).round(3)*100
    )
```

```python
pd.crosstab(
    df["Cholesterol"].eq(0),
    df["ExerciseAngina"],
    normalize="index"
).round(3) * 100
```
